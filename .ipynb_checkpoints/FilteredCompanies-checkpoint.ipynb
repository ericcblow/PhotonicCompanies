# Filter Companies by Country, U.S. Region, or U.S. State

This notebook filters `Photonic Companies.xlsx` by location in column F across all sheets and writes `FilteredCompanies.xlsx`.


```python
from pathlib import Path

# -----------------------------------------------------------------------------
# User configuration
# -----------------------------------------------------------------------------
INPUT_EXCEL = Path("Photonic Companies.xlsx")
OUTPUT_EXCEL = Path("FilteredCompanies.xlsx")

# Edit these lists for the filters you want to retrieve.
# You may use any combination of countries, U.S. regions, and individual U.S. states.
# Examples:
#   COUNTRIES = ["Japan", "Germany"]
#   COUNTRIES = ["United States"]
#   US_REGIONS = ["West Coast"]
#   US_REGIONS = ["East Coast", "Midwest"]
#   US_STATES = ["California"]
#   US_STATES = ["California", "Massachusetts", "Texas"]
#   US_STATES = ["CA", "MA", "TX"]
#   COUNTRIES = ["Japan"] and US_STATES = ["California"]
COUNTRIES = []
US_REGIONS = []
US_STATES = ["California"]

# Workbook layout assumptions from the current database.
SHEET_NAMES = None          # None = process every sheet in the workbook; or set e.g. ["Startup", "Large Company"]
HEADER_ROW = 2              # Header row is row 2
DATA_START_ROW = 3          # Company data starts on row 3
LOCATION_COLUMN = "F"       # Location column
FIRST_DATA_COLUMN = "B"     # Company Name starts in column B; column A is blank in the current workbook
```

```python
def normalize_text(value):
    """Return a lowercase ASCII-ish string with punctuation separated by spaces."""
    if value is None:
        return ""
    text = str(value)
    text = unicodedata.normalize("NFKD", text)
    text = "".join(ch for ch in text if not unicodedata.combining(ch))
    text = text.lower()
    text = re.sub(r"[^a-z0-9]+", " ", text)
    return re.sub(r"\s+", " ", text).strip()


def build_country_terms(countries, alias_map=None):
    """Build normalized search terms for each user-requested country."""
    alias_map = alias_map or {}
    normalized_alias_map = {
        normalize_text(canonical): {normalize_text(canonical), *(normalize_text(x) for x in aliases)}
        for canonical, aliases in alias_map.items()
    }

    country_terms = {}
    for country in countries:
        country_norm = normalize_text(country)
        terms = {country_norm}

        # Direct canonical match.
        if country_norm in normalized_alias_map:
            terms.update(normalized_alias_map[country_norm])

        # Reverse lookup: user may type "USA" instead of "United States".
        for canonical_norm, aliases in normalized_alias_map.items():
            if country_norm == canonical_norm or country_norm in aliases:
                terms.add(canonical_norm)
                terms.update(aliases)

        country_terms[str(country)] = sorted(t for t in terms if t)

    return country_terms


def contains_normalized_term(location_norm, term_norm):
    """Word-boundary match on already-normalized text."""
    if not location_norm or not term_norm:
        return False
    pattern = rf"(?<![a-z0-9]){re.escape(term_norm)}(?![a-z0-9])"
    return re.search(pattern, location_norm) is not None


def matched_countries(location, country_terms):
    """Return the requested countries that match a location string."""
    location_norm = normalize_text(location)
    matches = []
    for country, terms in country_terms.items():
        if any(contains_normalized_term(location_norm, term) for term in terms):
            matches.append(country)
    return matches


def build_us_region_terms(us_regions, region_state_map=None, state_to_abbr=None):
    """Build normalized state-name/state-abbreviation terms for each requested U.S. region."""
    region_state_map = region_state_map or {}
    state_to_abbr = state_to_abbr or {}

    # Normalize region names for flexible lookup, e.g. "west coast" or "West Coast".
    normalized_region_lookup = {normalize_text(region): region for region in region_state_map}

    region_terms = {}
    for requested_region in us_regions:
        requested_norm = normalize_text(requested_region)
        canonical_region = normalized_region_lookup.get(requested_norm)
        if canonical_region is None:
            valid_regions = ", ".join(region_state_map.keys())
            raise ValueError(f"Unknown U.S. region: {requested_region!r}. Valid options are: {valid_regions}")

        terms = set()
        for state in region_state_map[canonical_region]:
            terms.add(normalize_text(state))
            abbr = state_to_abbr.get(state)
            if abbr:
                terms.add(normalize_text(abbr))

        region_terms[canonical_region] = sorted(t for t in terms if t)

    return region_terms


def matched_us_regions(location, region_terms):
    """Return the requested U.S. regions that match a location string."""
    location_norm = normalize_text(location)
    matches = []
    for region, terms in region_terms.items():
        if any(contains_normalized_term(location_norm, term) for term in terms):
            matches.append(region)
    return matches


def build_us_state_terms(us_states, state_to_abbr=None):
    """Build normalized full-name/abbreviation terms for each requested U.S. state."""
    state_to_abbr = state_to_abbr or {}
    abbr_to_state = {abbr.upper(): state for state, abbr in state_to_abbr.items()}

    normalized_state_lookup = {normalize_text(state): state for state in state_to_abbr}
    normalized_abbr_lookup = {normalize_text(abbr): state for abbr, state in abbr_to_state.items()}

    state_terms = {}
    for requested_state in us_states:
        requested_norm = normalize_text(requested_state)
        canonical_state = normalized_state_lookup.get(requested_norm) or normalized_abbr_lookup.get(requested_norm)
        if canonical_state is None:
            valid_states = ", ".join(state_to_abbr.keys())
            valid_abbrs = ", ".join(state_to_abbr.values())
            raise ValueError(
                f"Unknown U.S. state: {requested_state!r}. "
                f"Use a full state name ({valid_states}) or abbreviation ({valid_abbrs})."
            )

        terms = {normalize_text(canonical_state)}
        abbr = state_to_abbr.get(canonical_state)
        if abbr:
            terms.add(normalize_text(abbr))

        state_terms[canonical_state] = sorted(t for t in terms if t)

    return state_terms


def matched_us_states(location, state_terms):
    """Return the requested U.S. states that match a location string."""
    location_norm = normalize_text(location)
    matches = []
    for state, terms in state_terms.items():
        if any(contains_normalized_term(location_norm, term) for term in terms):
            matches.append(state)
    return matches


def make_unique_sheet_name(workbook, base_name):
    """Excel sheet names must be unique and <=31 characters."""
    safe = re.sub(r"[\/*?:\[\]]", "_", str(base_name)).strip() or "Sheet"
    safe = safe[:31]
    existing = set(workbook.sheetnames)
    if safe not in existing:
        return safe
    for idx in range(1, 1000):
        suffix = f"_{idx}"
        candidate = safe[:31-len(suffix)] + suffix
        if candidate not in existing:
            return candidate
    raise RuntimeError(f"Could not create a unique sheet name for {base_name!r}")


def style_worksheet(ws):
    """Apply lightweight formatting for readability."""
    if ws.max_row < 1:
        return

    header_fill = PatternFill("solid", fgColor="1F4E78")
    header_font = Font(color="FFFFFF", bold=True)
    thin_gray = Side(style="thin", color="D9E2F3")
    border = Border(left=thin_gray, right=thin_gray, top=thin_gray, bottom=thin_gray)

    for cell in ws[1]:
        cell.fill = header_fill
        cell.font = header_font
        cell.alignment = Alignment(horizontal="center", vertical="center", wrap_text=True)
        cell.border = border

    for row in ws.iter_rows(min_row=2, max_row=ws.max_row):
        for cell in row:
            cell.alignment = Alignment(vertical="top", wrap_text=True)
            cell.border = border

    ws.freeze_panes = "A2"
    ws.auto_filter.ref = ws.dimensions

    for column_cells in ws.columns:
        column_letter = get_column_letter(column_cells[0].column)
        max_length = 0
        for cell in column_cells:
            if cell.value is not None:
                max_length = max(max_length, len(str(cell.value)))
        ws.column_dimensions[column_letter].width = min(max(max_length + 2, 12), 55)


# Excel Tables are intentionally not added.
# A sheet-level AutoFilter is applied in style_worksheet().
# This keeps the generated .xlsx simple and avoids Excel repair warnings.
```

```python
def filter_companies_by_location(
    input_excel=None,
    output_excel=None,
    countries=None,
    us_regions=None,
    us_states=None,
    sheet_names=None,
    header_row=None,
    data_start_row=None,
    location_column=None,
    first_data_column=None,
    country_aliases=None,
    us_region_states=None,
    us_state_to_abbr=None,
):
    """
    Filter company rows by country, U.S. region, and/or individual U.S. state,
    then write a new workbook.

    Output workbook contents:
    - FilteredCompanies: combined list across all processed sheets.
    - One sheet per original source sheet containing only matched rows from that sheet.

    Notes:
    - Arguments default to the latest values in the configuration cells at call time.
    - This avoids the common notebook issue where old default arguments persist after
      editing `COUNTRIES`, `US_REGIONS`, or `US_STATES`.
    """
    if input_excel is None:
        input_excel = INPUT_EXCEL
    if output_excel is None:
        output_excel = OUTPUT_EXCEL
    if countries is None:
        countries = COUNTRIES
    if us_regions is None:
        us_regions = US_REGIONS
    if us_states is None:
        us_states = US_STATES
    if sheet_names is None:
        sheet_names = SHEET_NAMES
    if header_row is None:
        header_row = HEADER_ROW
    if data_start_row is None:
        data_start_row = DATA_START_ROW
    if location_column is None:
        location_column = LOCATION_COLUMN
    if first_data_column is None:
        first_data_column = FIRST_DATA_COLUMN
    if country_aliases is None:
        country_aliases = COUNTRY_ALIASES
    if us_region_states is None:
        us_region_states = US_REGION_STATES
    if us_state_to_abbr is None:
        us_state_to_abbr = US_STATE_TO_ABBR

    input_excel = Path(input_excel)
    output_excel = Path(output_excel)
    countries = countries or []
    us_regions = us_regions or []
    us_states = us_states or []

    if not input_excel.exists():
        raise FileNotFoundError(f"Input workbook not found: {input_excel.resolve()}")
    if not countries and not us_regions and not us_states:
        raise ValueError("Please set COUNTRIES, US_REGIONS, and/or US_STATES to at least one filter.")

    source_wb = load_workbook(input_excel, data_only=False)
    selected_sheets = sheet_names or source_wb.sheetnames
    missing = [name for name in selected_sheets if name not in source_wb.sheetnames]
    if missing:
        raise ValueError(f"These requested sheets were not found: {missing}")

    first_col_idx = column_index_from_string(first_data_column)
    location_col_idx = column_index_from_string(location_column)
    country_terms = build_country_terms(countries, country_aliases)
    region_terms = build_us_region_terms(us_regions, us_region_states, us_state_to_abbr)
    state_terms = build_us_state_terms(us_states, us_state_to_abbr)

    # Use one consistent source-column span across all sheets. This avoids creating
    # extra data columns in the combined sheet without a corresponding header.
    global_last_col_idx = max(source_wb[name].max_column for name in selected_sheets)
    global_headers = []
    for col_idx in range(first_col_idx, global_last_col_idx + 1):
        header_value = None
        for sheet_name in selected_sheets:
            candidate = source_wb[sheet_name].cell(header_row, col_idx).value
            if candidate is not None and str(candidate).strip():
                header_value = candidate
                break
        global_headers.append(header_value if header_value is not None else f"Column {get_column_letter(col_idx)}")

    output_wb = Workbook()
    output_wb.remove(output_wb.active)

    combined_ws = output_wb.create_sheet("FilteredCompanies")
    combined_header_written = False
    summary = defaultdict(int)
    total_matches = 0

    for source_sheet_name in selected_sheets:
        source_ws = source_wb[source_sheet_name]
        last_col_idx = global_last_col_idx
        headers = list(global_headers)

        sheet_output_name = make_unique_sheet_name(output_wb, source_sheet_name)
        filtered_ws = output_wb.create_sheet(sheet_output_name)
        per_sheet_header = [
            "Source Row",
            "Matched Filters",
            "Matched Countries",
            "Matched U.S. Regions",
            "Matched U.S. States",
            *headers,
        ]
        filtered_ws.append(per_sheet_header)

        if not combined_header_written:
            combined_ws.append([
                "Source Sheet",
                "Source Row",
                "Matched Filters",
                "Matched Countries",
                "Matched U.S. Regions",
                "Matched U.S. States",
                *headers,
            ])
            combined_header_written = True

        for row_idx in range(data_start_row, source_ws.max_row + 1):
            location_value = source_ws.cell(row_idx, location_col_idx).value
            country_matches = matched_countries(location_value, country_terms)
            region_matches = matched_us_regions(location_value, region_terms)
            state_matches = matched_us_states(location_value, state_terms)
            if not country_matches and not region_matches and not state_matches:
                continue

            row_values = [source_ws.cell(row_idx, col).value for col in range(first_col_idx, last_col_idx + 1)]
            matched_country_text = ", ".join(country_matches)
            matched_region_text = ", ".join(region_matches)
            matched_state_text = ", ".join(state_matches)
            matched_filter_text = ", ".join([*country_matches, *region_matches, *state_matches])

            filtered_ws.append([row_idx, matched_filter_text, matched_country_text, matched_region_text, matched_state_text, *row_values])
            combined_ws.append([source_sheet_name, row_idx, matched_filter_text, matched_country_text, matched_region_text, matched_state_text, *row_values])

            total_matches += 1
            summary[source_sheet_name] += 1

    # If no matches exist, keep a readable output workbook instead of failing silently.
    if total_matches == 0:
        combined_ws.append(["No matching companies found", None, ", ".join([*countries, *us_regions, *us_states])])

    # Style sheets. Use a normal sheet AutoFilter instead of Excel Tables;
    # this avoids Excel repair warnings caused by duplicate table/sheet filters.
    for ws in output_wb.worksheets:
        style_worksheet(ws)

    output_wb.save(output_excel)

    print(f"Saved: {output_excel.resolve()}")
    print(f"Countries searched: {countries}")
    print(f"U.S. regions searched: {us_regions}")
    print(f"U.S. states searched: {us_states}")
    print(f"Total matching rows: {total_matches}")
    for sheet_name in selected_sheets:
        print(f"  {sheet_name}: {summary[sheet_name]}")

    return output_excel
```

```python
# Backward-compatible wrapper for older run cells.
# You can call either filter_companies_by_location() or the older name
# filter_companies_by_country(); both now support COUNTRIES, US_REGIONS, and US_STATES.
def filter_companies_by_country(*args, **kwargs):
    return filter_companies_by_location(*args, **kwargs)
```

```python
# Run the filter. Edit COUNTRIES, US_REGIONS, and/or US_STATES in the configuration cell, then rerun this cell.
filtered_file = filter_companies_by_location(
    countries=COUNTRIES,
    us_regions=US_REGIONS,
    us_states=US_STATES,
)
filtered_file
```
